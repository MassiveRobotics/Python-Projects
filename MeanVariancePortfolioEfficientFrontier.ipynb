```python
import numpy as np

# portfolio weights (known weights)

weights =np.array([ 0.40, 0.35, 0.25 ])

# expected annual returns

expected_returns = np.array([ 0.10, 0.08, 0.12 ])

# covariance matrix

cov_matrix = np.array([
    [.0400, .0120, .0080],
    [.0120, .2250, .0090],
    [.0080, .0090, .0625]
])

# expected portfolio return

portfolio_return = weights @ expected_returns

# portfolio variance

portfolio_variance = weights.T @ cov_matrix @ weights

# portfolio standard deviation

portfolio_volatility = np.sqrt(portfolio_variance)

print(f"Expected Portfolio Return: {portfolio_return:.2%}")
print(f"Portfolio Variance: {portfolio_variance:.6f}")
print(f"Portfolio Volatility: {portfolio_volatility:.2%}")
```

```python
import numpy as np
from scipy.optimize import minimize

expected_returns = np.array([0.10, 0.08, 0.12])

cov_matrix = np.array([
    [.0400, .0120, .0080],
    [.0120, .2250, .0090],
    [.0080, .0090, .0625]
])

def portfolio_variance(weights):
  return weights.T @ cov_matrix @ weights

target_return = 0.10

constraints = {
    "type": "eq",
    "fun": lambda weights: np.sum(weights) - 1
}

bounds = (
    (0, 1),
    (0, 1),
    (0, 1)
)

initial_weights = np.array([0.33, 0.33, 0.33])

result = minimize(
    portfolio_variance,
    initial_weights,
    method="SLSQP",
    constraints=constraints,
    bounds=bounds
)

optimal_weights = result.x

optimal_return = optimal_weights @ expected_returns
optimal_variance = portfolio_variance(optimal_weights)
optimal_volatility = np.sqrt(optimal_variance)

print(f"Optimal Portfolio Weights: {optimal_weights}")
print(f"Optimal Portfolio Return: {optimal_return:.2%}")
print(f"Optimal Portfolio Variance: {optimal_variance:.6f}")
print(f"Optimal Portfolio Volatility: {optimal_volatility:.2%}")
```
