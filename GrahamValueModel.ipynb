```python
def analyze_graham_value(ticker,price,eps,book_value):

  if eps <= 0 or book_value <= 0:
    return f"{ticker}: fails criteria"

  graham_number = (22.5 * eps * book_value) ** .5
  if price < graham_number:
    discount = ((graham_number-price) / graham_number) * 100
    return f"{ticker}: underpriced"
  else:
      premium =((price-graham_number) / graham_number) * 100
      return f"{ticker}: overpriced"

portfolio_sandbox = [{"ticker": "comp a", "price": 45.00, "eps": 3.50, "book_value": 22.00 },{"ticker": "comp b", "price": 120.00, "eps": 2.10, "book_value": 15.50}, {"ticker": "comp c","price": 15, "eps": -.5, "book_value": 8.00}]

print("running graham filter")
for stock in portfolio_sandbox:
    result = analyze_graham_value(
            stock["ticker"],
            stock["price"],
            stock["eps"],
            stock["book_value"])
print(result)

```
